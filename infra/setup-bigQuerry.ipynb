```python
from google.cloud import bigquery

project_id = "ai-dj-487610"
dataset_id = "Spotify_Tracks"

client = bigquery.Client(project=project_id)

dataset = bigquery.Dataset(f"{project_id}.{dataset_id}")
dataset.location = "europe-west1"

# Use exists_ok=True to make this command safe to run multiple times
dataset = client.create_dataset(dataset, exists_ok=True)
print(f"Created dataset {dataset.project}.{dataset.dataset_id}")
```

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "score": [95, 87, 92]
})

table_id = f"{project_id}.{dataset_id}.tracks"
job = client.load_table_from_dataframe(df, table_id)
job.result()

print(f"Loaded {len(df)} rows into {table_id}")
```
